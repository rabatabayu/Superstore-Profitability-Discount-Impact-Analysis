```python
import pandas as pd
```

```python
# Import Data
df = pd.read_csv("Sample - Superstore.csv", encoding="windows-1252")
df #Display Data
```

```python
# Pengecekan Data Awal
df.info()
```

```python
# Check Missing Value
df.isnull().sum()
```

```python
# Check Duplicate
df.duplicated().sum()
```

```python
df[["Order Date", "Ship Date"]].head(30)
```

```python
# Ubah Format Date
df["Order Date"] = pd.to_datetime(df["Order Date"], format="%m/%d/%Y") 
df["Ship Date"] = pd.to_datetime(df["Ship Date"], format="%m/%d/%Y") 
```

```python
df.info()
```

```python
# Check unique values in categorical columns

print("Ship Mode:")
print(df["Ship Mode"].value_counts())

print("\nSegment:")
print(df["Segment"].value_counts())

print("\nRegion:")
print(df["Region"].value_counts())

print("\nCategory:")
print(df["Category"].value_counts())

print("\nSub-Category:")
print(df["Sub-Category"].value_counts())
```

```python
# Check numerical data distribution
df[["Sales", "Quantity", "Discount", "Profit"]].describe()
```

```python
# Check invalid numerical values
print("Sales <= 0:", (df["Sales"] <= 0).sum())
print("Quantity <= 0:", (df["Quantity"] <= 0).sum())
print("Discount < 0:", (df["Discount"] < 0).sum())
print("Discount > 1:", (df["Discount"] > 1).sum())
```

```python
# Check invalid shipping date
print("Ship Date before Order Date:",
      (df["Ship Date"] < df["Order Date"]).sum())
```

```python
for col in ["Sales", "Profit"]:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    print(col)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Total Outlier:", len(outlier))
    print()
```

```python
# Show transactions with the lowest profit
df.sort_values("Profit").head(10)[
    ["Category", "Sub-Category", "Sales", "Discount", "Profit"]]
```

```python
# Check categorical frequency

categorical_cols = [
    "Ship Mode",
    "Segment",
    "Region",
    "Category",
    "Sub-Category"
]

for col in categorical_cols:
    print(f"\n{col}")
    print(df[col].value_counts())
```

```python
# Check rare categories (<1%)

for col in categorical_cols:
    proportion = df[col].value_counts(normalize=True) * 100
    rare = proportion[proportion < 1]

    print(f"\nRare categories in {col}:")
    print(rare)
```

```python
# Export cleaned dataset
df.to_csv(
    "Superstore_Cleaned.csv",
    index=False,
    encoding="utf-8-sig"
)
```
